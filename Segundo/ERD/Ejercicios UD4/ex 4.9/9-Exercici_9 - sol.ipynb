# Detección de impago en microcréditos

Importar librerías

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
n=2500
df = pd.DataFrame({
    "ingresos_mensuales": np.random.normal(1200,400,n).clip(400,4000),
    "deuda_actual": np.random.normal(600,300,n).clip(0,3000),
    "historial_crediticio": np.random.randint(1,10,n),
    "edad": np.random.randint(18,70,n),
    "num_prestamos_previos": np.random.poisson(2,n)
})
prob = 0.5*(df["deuda_actual"]>1500)+0.3*(df["historial_crediticio"]<4)
df["IMPAGO"]=np.random.binomial(1,np.clip(prob,0,1))
df.to_csv("microcreditos_knn.csv", index=False)
```

Cargar datos

```python
df = pd.read_csv("microcreditos_knn.csv")
df.head()
```

EDA

```python
df.describe()
sns.pairplot(df, hue="IMPAGO")
```

Separar variables

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = df.drop("IMPAGO", axis=1)
y = df["IMPAGO"]
X_scaled = scaler.fit_transform(X)
```

Escalar

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

Train/Test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

Modelo inicial

Perfil de usuarios

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

Evaluación

```python
from sklearn.metrics import classification_report
pred = knn.predict(X_test)
print(classification_report(y_test, pred))
```

Elegir K óptimo

```python
error_rate = []
for k in range(1,40):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    pred_k = knn.predict(X_test)
    error_rate.append(np.mean(pred_k != y_test))

plt.plot(range(1,40), error_rate, marker='o')
plt.show()
```
